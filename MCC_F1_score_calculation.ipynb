```python
def get_metrics(df, feature_indices):
    X = df.iloc[:, feature_indices]
    target = df.columns[-1]
    y = df[target]

    # Split data into train and test
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
    
    scaler = StandardScaler()
    train_x_sc = scaler.fit_transform(X_train)
    test_x_sc = scaler.transform(X_test)

    model = RandomForestClassifier(n_estimators=100, random_state=42)

    model.fit(train_x_sc, y_train)
    y_pred = model.predict(test_x_sc)
    accuracy = accuracy_score(y_test, y_pred)
    
    f1 = f1_score(y_test, y_pred, average='weighted')
    mcc = matthews_corrcoef(y_test, y_pred)
    
    return {"accuracy":accuracy, "f1":f1, "mcc":mcc}
```

```python

import os

def save_results(algo, features, datasets = datasets, save_dir="./MCC_F1_results/high_dim"):
    result = {'bal':{}, 'imbal':{}}

    for i,dataset in enumerate(datasets): 
        path_bal = f"{bal_path}/{dataset}"
        path_imbal = f"{imbal_path}/{dataset}"

        try:
            bal = False
            if(os.path.exists(path_bal)):
                data = pd.read_csv(path_bal)
                bal = True
            else:
                data = pd.read_csv(path_imbal)

            metrics = get_metrics(data, features[i])
            if(bal):
                result['bal'][dataset] = metrics
            else:
                result['imbal'][dataset] = metrics
            
            print(f"Processed {dataset}: {metrics}")
        except Exception as e: 
            print(f"Error reading {dataset}: {e}")
            continue

    with open(f"{save_dir}/{algo}.json", "w") as f:
        json.dump(result, f, indent=4)
```
