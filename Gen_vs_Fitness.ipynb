```python
import os
import json
import matplotlib.pyplot as plt

# Base directory containing all parameter-specific folders
base_dir = r"PM_NoEM_corrected"

# Dictionary to hold fitness values for each folder
fitness_data = {}

# Loop over all subfolders
for folder in os.listdir(base_dir):
    folder_path = os.path.join(base_dir, folder)
    json_path = os.path.join(folder_path, "SIM", "RUN", "gen_fitness.json")
    
    if os.path.isdir(folder_path) and os.path.exists(json_path):
        with open(json_path, "r") as f:
            data = json.load(f)
            # Extract the fitness values (second element in each sublist under 'bests')
            fitness_values = [entry[1] for entry in data.get("bests", [])]
            folder = folder.replace("LOGS_LD_", "").replace("pm", "p").replace("_"," ").replace("a","α").replace("b","β") #.replace("_", ", ")
            fitness_data[folder] = fitness_values

# Plotting
plt.figure(figsize=(12,6))
for param, values in sorted(fitness_data.items()):
    plt.plot(values, marker='o', label=param)

plt.title("Best Fitness over Generations for Different Parameters")
plt.xlabel("Generation")
plt.ylabel("Fitness")
plt.legend(title="Configurations", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
